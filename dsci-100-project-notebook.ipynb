```r
basketball_avg_table <- basketball_tidy|>
# I used summarize to apply mean, max, or min to the different variables and assigned them all new and descriptive
# column names so that I can easily tell what each column represents
    summarize(
        year_mean = mean(year, na.rm = TRUE), 
        year_max = max(year, na.rm = TRUE),
        year_min = min(year, na.rm = TRUE),
        seed_mean = mean(seed, na.rm = TRUE),
        seed_max = max(seed, na.rm = TRUE),
        seed_min = min(seed, na.rm = TRUE),
    )

basketball_avg_table
```

```r
# this plot is going to show the distribution of teams within each seed value (1-16)

seed_plot <- ggplot(basketball_tidy, aes(x = seed))+
    geom_histogram(binwidth = 1, color = 'white')+
    scale_x_continuous(breaks = 0:16)+
    labs(x= 'Seed Number (1-16)', y= 'Number of Teams', title= 'Distribution of Teams with each Seed Number')
seed_plot
```

```r
# in this cell i will calculate the proportions of each bid type and put them in a new column to make my plotting in the next
# cell easier
bid_proportions <- basketball_tidy|>
    group_by(bid)|>
    summarize(count = n())|>
    mutate(proportion = count/sum(count))
```

```r
#
bid_plot <- ggplot(bid_proportions, aes(x = bid, y = proportion))+
    geom_bar(stat = 'identity') +
    labs(x= 'Bid Type', y = 'Proportion of Teams', title = 'Proportion of Auto-Bid vs. At Large Teams')

bid_plot
```
